## Part A

````python
import pandas as pd
# error1 | Type: Import Error | Wrong Code: Import pandas as pd | Reason: The Library name is pandas, not panda | Fix: Change panda to pandas
````

## Part B

````python
# ERROR 2 | Type: FileNotFoundError | Wrong code: "Flipkart_Order_10.xlsx" | Reason: The uploaded Excel file has a different filename. | FIX: Use the uploaded DQ1_Datasheet.xlsx file.
# ERROR 3 | Type: SheetNameError | Wrong code: sheet_name="Order" | Reason: The Excel sheet is named "Orders", not "Order". | FIX: Changed the sheet name to "Orders".
df = pd.read_excel("DQ1_Datasheet.xlsx", sheet_name="Orders")
````

## Part C

````python
print(df.head())

# ERROR 4 | Type: Syntax/Attribute Error | Wrong code: df.shape() | Reason: shape is an attribute, not a function, so parentheses should not be used. | FIX: Changed df.shape() to df.shape.
print("Rows and columns:", df.shape)

# ERROR 5 | Type: Attribute Error | Wrong code: df.column | Reason: The pandas DataFrame attribute is "columns", not "column". | FIX: Changed df.column to df.columns.
print("Columns:", list(df.columns))
````

## Part D

````python
# ERROR 6 | Type: KeyError | Wrong code: df["price"] | Reason: The actual column name is "Price" with a capital P. | FIX: Changed price to Price.
df["Amount"] = df["Price"] * df["Quantity"]

# ERROR 7 | Type: Calculation/Logic Error | Wrong code: df["Price"] + df["Quantity"] | Reason: Total amount must be Price per item multiplied by Quantity, not added. | FIX: Changed + to *.
df["Amount"] = df["Price"] * df["Quantity"]

df["Total"] = df["Amount"] + df["Amount"] * 18 / 100
df["Boxes"] = df["Quantity"] // 3
df["Left"] = df["Quantity"] % 3

print(df[["Order_ID", "Amount", "Total", "Boxes", "Left"]])
````

## Part E

````python
slow_orders = (df["Delivery_Days"] > 5).sum()

# ERROR 8 | Type: Logic Error | Wrong code: (df["Rating"] > 4).sum() | Reason: The question asks for rating 4 or more, so rating 4 must also be counted. | FIX: Changed > 4 to >= 4.
good_ratings = (df["Rating"] >= 4).sum()

print("Slow deliveries (more than 5 days):", slow_orders)
print("Orders with rating 4 or more     :", good_ratings)
````

## Part F

````python
# ERROR 9 | Type: Syntax Error | Wrong code: elif days <= 5 | Reason: An elif statement must end with a colon. | FIX: Added : after the condition.
# ERROR 10 | Type: Logic Error | Wrong code: if rating >= 2 | Reason: For a slow delivery, a low rating of 2 or below should trigger a stronger customer response. | FIX: Changed >= 2 to <= 2.

for i in range(len(df)):
    order_id = df["Order_ID"][i]
    days = df["Delivery_Days"][i]
    rating = df["Rating"][i]

    if days <= 2:
        status = "Fast"

    elif days <= 5:
        status = "Normal"
    else:
        status = "Slow"

    if days > 5:

        if rating <= 2:
            action = "Call customer"
        else:
            action = "Send apology SMS"
    else:
        action = "No action"

    print(order_id, "|", status, "| Rating", rating, "|", action)
````
